```python
import numpy as np

min_yield = df['Yield'].min()  # 0.0773
max_yield = df['Yield'].max()  # 67.7621

# Define bins within min-max range
bins = [min_yield - 1, 20, 40, max_yield + 1]  # 3 categories: Low, Medium, High

# Categorize Yield
df['Yield_category'] = pd.cut(df['Yield'], bins=bins, labels=['Low', 'Medium', 'High'], include_lowest=True)
```

```python
season_cols = ['Season_Autumn', 'Season_Kharif', 'Season_Rabi', 'Season_Summer',
               'Season_Whole Year', 'Season_Winter']
df['Season'] = df[season_cols].idxmax(axis=1).str.replace('Season_', '')
```

```python
from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(df['Season'], df['Yield_category'])
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("p-value:", p)
if p < 0.05:
    print("Significant association exists between Season and Yield_category")
else:
    print("No significant association between Season and Yield_category")
```
